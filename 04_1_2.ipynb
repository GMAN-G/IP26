```python
gBin = np.zeros((RGB_img.shape[0], RGB_img.shape[1]), np.uint8)
```

```python
for h in range(RGB_img.shape[0]):
  for w in range(RGB_img.shape[1]):
    if(np.int32(RGB_img[h,w][0])<140):
      gBin[h,w]=255
    else:
      gBin[h,w]=0

cv2_imshow(gBin)
```
