```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
img = cv2.imread('/content/gdrive/My Drive/Colab26/Images/lena256.jpg')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

gray_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
print(gray_img.shape)
cv2_imshow(gray_img)
```

```python
meanimg = cv2.blur(gray_img, (5, 5))
cv2_imshow(meanimg)
```

```python
edgeImg = cv2.addWeighted(gray_img, 1.0, meanimg, -1.0, 0)
cv2_imshow(edgeImg)
```

```python
unsharpImg = cv2.addWeighted(gray_img, 1.0, edgeImg, 2.0, 0)
cv2_imshow(unsharpImg)
```
