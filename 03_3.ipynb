```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
def saturation(value):
  if(value > 255):
    value = 255
  elif(value < 0):
    value = 0
  return value
```

```python
img = cv2.imread('/content/gdrive/My Drive/Colab26/Images/aero2.bmp')
gray_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

cv2_imshow(gray_img)
```

```python
gray_Out = np.zeros((gray_img.shape[0], gray_img.shape[1]), dtype=np.ubyte)
```

```python
min = 256
max = 0

for h in range(gray_img.shape[0]):
  for w in range(gray_img.shape[1]):
    if gray_img[h,w] < min:
      min = gray_img[h, w]
    if gray_img[h,w] > max:
      max = gray_img[h, w]

print(min)
print(max)
```

```python
for h in range(gray_img.shape[0]):
  for w in range(gray_img.shape[1]):
    val = (gray_img[h,w] - min) / (max - min) * 255
    gray_Out[h, w] = val

cv2_imshow(gray_Out)
```

```python
img = cv2.imread('/content/gdrive/My Drive/Colab26/Images/girl.jpg')
gray_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

cv2_imshow(gray_img)
```

```python
gray_Mul = np.zeros((gray_img.shape[0], gray_img.shape[1]), dtype=np.ubyte)
```

```python
mulVal = 4

for h in range(gray_img.shape[0]):
  for w in range(gray_img.shape[1]):
    val = np.int32(gray_img[h, w]) * mulVal
    gray_Mul[h, w] = saturation(val)

cv2_imshow(gray_Mul)
```

```python
gray_Div = np.zeros((gray_img.shape[0], gray_img.shape[1]), dtype=np.ubyte)
```

```python
divVal = 4

for h in range(gray_img.shape[0]):
  for w in range(gray_img.shape[1]):
    val = np.int32(gray_Mul[h, w]) / divVal
    gray_Div[h, w] = saturation(val)

cv2_imshow(gray_Div)
```
