```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
img = cv2.imread('/content/gdrive/My Drive/Colab26/Images/girl.jpg')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray_img = cv2.cvtColor(RGB_img, cv2.COLOR_RGB2GRAY)

cv2_imshow(gray_img)
```

```python
gray_Plus = np.zeros((gray_img.shape[0], gray_img.shape[1]), dtype=np.ubyte)
```

```python
val = 140

for h in range(gray_img.shape[0]):
  for w in range(gray_img.shape[1]):
    plusVal = np.int32(gray_img[h][w]) + val
    if(plusVal > 255):
      plusVal = 255
    gray_Plus[h][w] = plusVal

cv2_imshow(gray_Plus)
```

```python
gray_Origin = np.zeros((gray_img.shape[0], gray_img.shape[1]), dtype=np.ubyte)
```

```python
val = 140

for h in range(gray_img.shape[0]):
  for w in range(gray_img.shape[1]):
    plusVal = np.int32(gray_Plus[h][w]) - val
    if(plusVal > 255):
      plusVal = 255
    gray_Origin[h][w] = plusVal

cv2_imshow(gray_Origin)
```

```python
img = cv2.imread('/content/gdrive/My Drive/Colab26/Images/aero2.bmp')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray_img1 = cv2.cvtColor(RGB_img, cv2.COLOR_RGB2GRAY)

cv2_imshow(gray_img1)
print(gray_img1.shape)
```

```python
img = cv2.imread('/content/gdrive/My Drive/Colab26/Images/milkdrop.bmp')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray_img2 = cv2.cvtColor(RGB_img, cv2.COLOR_RGB2GRAY)

cv2_imshow(gray_img2)
print(gray_img2.shape)
```

```python
gray_IPlus = np.zeros((gray_img.shape[0], gray_img.shape[1]), dtype=np.ubyte)
```

```python
for h in range(gray_img1.shape[0]):
  for w in range(gray_img1.shape[1]):
    plusVal = np.int32(gray_img1[h][w]) + np.int32(gray_img2[h][w])
    plusVal = plusVal / 2
    if(plusVal > 255):
      plusVal = 255
    gray_IPlus[h][w] = plusVal

cv2_imshow(gray_IPlus)
```

```python
weight = 0.1

for h in range(gray_img1.shape[0]):
  for w in range(gray_img1.shape[1]):
    plusVal = np.int32(gray_img1[h][w]) * (1-weight) + np.int32(gray_img2[h][w]) * weight
    if(plusVal > 255):
      plusVal = 255
    gray_IPlus[h][w] = plusVal

cv2_imshow(gray_IPlus)
```
