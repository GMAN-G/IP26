```python
img = cv2.imread('/content/gdrive/My Drive/Colab26/Images/lena256.jpg')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

gray_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
print(gray_img.shape)
cv2_imshow(gray_img)
```

```python
from random import randint
```

```python
noiseimg = gray_img.copy()
noiseRatio = 10;

for h in range(0, gray_img.shape[0]):
  for w in range(0, gray_img.shape[1]):
    val = randint(0, 100)
    if(val < noiseRatio / 2):
      noiseimg[h][w] = 255
    elif(val < noiseRatio):
      noiseimg[h][w] = 0

cv2_imshow(noiseimg)
```

```python
outimg = cv2.medianBlur(noiseimg, 3)
cv2_imshow(outimg)
```
