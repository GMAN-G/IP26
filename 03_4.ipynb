```python
def saturation(value):
  if(value > 255):
    value = 255
  elif(value < 0):
    value = 0
  return value
```

```python
img = cv2.imread('/content/gdrive/My Drive/Colab26/Images/lena256.jpg')
gray_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

cv2_imshow(gray_img)
```

```python
gray_Out = np.zeros((gray_img.shape[0], gray_img.shape[1]), dtype=np.ubyte)
```

```python
gray_Out = np.zeros((gray_img.shape[0], gray_img.shape[1]), dtype=np.ubyte)
planeNumber = 1
planeValue = 1

for i in range(planeNumber-1):
  planeValue = planeValue * 2
```

```python
for h in range(gray_img.shape[0]):
  for w in range(gray_img.shape[1]):
    imVal = np.int32(gray_img[h,w] / planeValue);
    if(imVal % 2 == 0):
      gray_Out[h,w] = 0
    else:
      gray_Out[h,w] = 255

cv2_imshow(gray_Out)
```
